```python
import pickle as pkl
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
import os 
```

```python


def txt2df(path):

    user_sequences = defaultdict(list)

    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split()
            user_id = int(parts[0])
            item_ids = list(map(int, parts[1:]))

            # append in order
            user_sequences[user_id].extend(item_ids)

    rows = []
    for user_id, items in user_sequences.items():
        for ts, item_id in enumerate(items):
            rows.append({
                "uid": user_id,
                "item_id": item_id+1,
                "timestamp": ts
            })

    df = pd.DataFrame(rows)
    return df
```

```python
def txt2df2(path):

    rows = []

    with open(path, "r") as f:
        for line in f:
            parts = line.strip().split()
            user_id = int(parts[0])
            item_ids = list(map(lambda x: int(x)+1, parts[1:]))
            
            rows.append({
                "uid": user_id,
                "item_id": item_ids,
            })
            

    df = pd.DataFrame(rows)
    return df
```

```python
from pathlib import Path
for domain in os.listdir('.'):
    path = Path(domain)
    if path.is_dir() and domain[0].isupper():
        print(domain)
        tgt_path = os.path.join('/home/jovyan/Samra/fl-recsys-adaptation/data/FedData/', domain)
        os.makedirs(tgt_path, exist_ok=True)
        txt2df2(path / 'train_data.txt').to_parquet(tgt_path + '/train_.parquet')
        txt2df2(path / 'valid_data.txt').to_parquet(tgt_path + '/val_.parquet')
        txt2df2(path / 'test_data.txt').to_parquet(tgt_path + '/test_.parquet')

        n_items = int(open(path / 'num_items.txt').read())
        item_dict = {k:k for k in range(n_items)}
        pkl.dump(item_dict, open(tgt_path + '/item_id_to_idx.pkl', 'wb'))
```
